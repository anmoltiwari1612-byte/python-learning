### 1. Introduction to Matplotlib
Matplotlib is a powerful library for creating visualizations in Python. To use it, we usually import the `pyplot` module as `plt`. We also need `pandas` to read our data file.

```python
import matplotlib.pyplot as plt
import pandas as pd

# Load the dataset
df = pd.read_csv('/content/sample_data/california_housing_train.csv')

# Show the first few rows to see what columns we have
display(df.head())
```

### 2. Basic Scatter Plot
A scatter plot helps us see the relationship between two variables. Let's look at how **Longitude** and **Latitude** relate (this effectively plots a map of the houses).

```python
plt.figure(figsize=(10, 6))

# plt.scatter(x, y) creates the plot
plt.scatter(df['longitude'], df['latitude'], alpha= 0.3)

# Adding labels and a title
plt.title('California Housing Locations')
plt.xlabel('Longitude')
plt.ylabel('Latitude')

# Show the plot
plt.show()
```

### 3. Basic Histogram
A histogram shows the distribution of a single variable. Let's see the distribution of **Median House Value**.

```python
plt.figure(figsize=(8, 5))

# plt.hist(data) creates the histogram
# 'bins' defines how many bars to use
plt.hist(df['median_house_value'], bins=50, color='skyblue', edgecolor='black')

plt.title('Distribution of Median House Values')
plt.xlabel('House Value')
plt.ylabel('Frequency')

plt.show()
```

### 4. Basic Bar Chart
Bar charts are used to compare values across different categories. Let's group the houses by their age and see the average number of rooms.

```python
plt.figure(figsize=(12, 6))

# We calculate the average rooms for each housing age
avg_rooms_by_age = df.groupby('housing_median_age')['total_rooms'].mean().head(20)

print(avg_rooms_by_age)
# plt.bar(categories, values) creates the bar chart
plt.bar(avg_rooms_by_age.index, avg_rooms_by_age.values, color='salmon')

plt.title('Average Rooms by Housing Median Age (First 20 Ages)')
plt.xlabel('Housing Median Age')
plt.ylabel('Average Total Rooms')

plt.show()
```

### 5. Box Plot
A box plot is excellent for seeing the 'spread' of data. It shows the median, the quartiles (where most data sits), and outliers (the dots outside the whiskers).

```python
plt.figure(figsize=(8, 6))

# plt.boxplot() shows the distribution of values
plt.boxplot(df['median_income'])

plt.title('Box Plot of Median Income')
plt.ylabel('Income (in tens of thousands)')
# Removing x-axis ticks for a single box plot
plt.xticks([1], ['All Houses'])

plt.show()
```

### 6. Correlation Heatmap
A heatmap uses colors to represent numbers. In this case, we'll calculate the 'correlation' between columns—which tells us how much one variable changes when another does.

*   **1.0 (Yellow/Bright)**: Perfect positive relationship.
*   **0.0 (Purple/Dark)**: No relationship.
*   **-1.0**: Perfect opposite relationship.

```python
plt.figure(figsize=(10, 8))

# Calculate correlation matrix
corr = df.corr()

# plt.imshow displays data as an image (heatmap)
plt.imshow(corr, cmap='viridis', interpolation='none')

# Adding a colorbar to show the scale
plt.colorbar()

# Adding labels to the axes
plt.xticks(range(len(corr.columns)), corr.columns, rotation=90)
plt.yticks(range(len(corr.columns)), corr.columns)

plt.title('Correlation Heatmap of Variables')
plt.show()
```
